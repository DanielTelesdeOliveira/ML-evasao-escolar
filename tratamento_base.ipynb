```python
#********************** BIBLIOTECAS UTILZIDAS ********************** 
import pandas as pd
import gdown
from dotenv import load_dotenv
import os

load_dotenv() #Carregando variáveis do ambiente
```

```python
#********************** Metodos e Variaveis Globas **********************
def ajustar_regiao(df, nome_coluna,id, nome_regiao): 
    df[nome_coluna] = df[nome_coluna].replace({id: nome_regiao})
    return df

def renomear_colunas(df, cols_antigas, cols_novas):
    mapeamento = dict(zip(cols_antigas, cols_novas))
    return df.rename(columns=mapeamento)


colunas_antigas = ['tx_resp_q08',            #Nome das colunas antes de alterar
                  'tx_resp_q09',
                  'tx_resp_q10c',
                  'tx_resp_q10e',
                  'tx_resp_q19',
                  'tx_resp_q20',
                  'tx_resp_q23i',
                  'tx_resp_q21e',
                  'tx_resp_q21d',
                  'tx_resp_q21c',
                  'tx_resp_q21a', 
                  'tx_resp_q23a', 
                  'tx_resp_q23b',
                  'tx_resp_q17',
                  'tx_resp_q14']
                

colunas_novas = ['Escolaridade_Mae',         #Nome das colunas apos alterar os nomes 
                 'Escolaridade_Pai',
                 'Incentivo_Estudos_Pais',
                 'Incentivo_Presenca_Pais',
                 'Reprovacao',
                 'Abandono_Escola',
                 'Incentivo_Estudos_Professores',
                 'Tempo_Lazer',
                 'Tempo_Trabalho_Fora', 
                 'Tempo_Trabalho_Casa',
                 'Tempo_Estudos_Fora', 
                 'Interesse_Ensino',
                 'Motivacao_Estudos',
                 'Idade_Inicio_Estudos', 
                 'Tempo_Viagem_Escola']
                 

variaveis_criticas = ['Escolaridade_Mae',         
                 'Escolaridade_Pai',
                 'Incentivo_Estudos_Pais',
                 'Incentivo_Presenca_Pais',
                 'Reprovacao',
                 'Abandono_Escola',
                 'Incentivo_Estudos_Professores',
                 'Tempo_Lazer',
                 'Tempo_Trabalho_Fora', 
                 'Tempo_Trabalho_Casa',
                 'Tempo_Estudos_Fora', 
                 'Interesse_Ensino',
                 'Motivacao_Estudos',
                 'Idade_Inicio_Estudos', 
                 'Tempo_Viagem_Escola']
```

```python
#********************** CARREGANDO BASE DE DADOS DA NUVEM **********************
id_arquivo = os.getenv("id_arquivo")
url = os.getenv("url")
url_arquivo = f'{url}{id_arquivo}'


output = 'base_inep.csv'    #Nome do arquivo CSV
gdown.download(url_arquivo, output, quiet=False)

df = pd.read_csv(output, sep=',', encoding='utf-8') #Criacao do primeiro datarame
```

```python
#********************** TRATAMENTO DOS DADOS **********************
df_EP = df[df['IN_PUBLICA'] == 0]   #Filtro para apenas o Ensino Publico
df_EP = ajustar_regiao(df_EP, 'ID_UF', 35, 'São Paulo')

df_EP.columns = df_EP.columns.str.lower()
df_EP = renomear_colunas(df_EP, colunas_antigas, colunas_novas)
df_EP = df_EP[variaveis_criticas]
```
